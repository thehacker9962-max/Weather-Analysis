```python
sns.histplot(df["Temp_C"], kde=True)
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
